STREAMING

it means we can stream the output from the llm while it is being generated

in traditional invoke method the model waits till the whole output is generatedd and then gives the output all at once 
but incase of streaming the model generates the output  as soon as some word is genrated hence improving the user experience

```python
#this is hw we do streaming

for chunk in model.stream("why do parrots speak"):
  print(chunk.text,end="",flush=True)


#end="" prevents Python from adding a newline after every chunk.
#flush=True displays each chunk immediately.
```

BATCHING


batching a collection of independent requests to a modelcan significantly improve he performance and reduce cost as the computing will be done in a parallel way

```python
responses=model.batch(["why are macaws majorely red?",
                      "why are penguins only seen in cold areas?",
                      "what is quantum computing"])

for response in responses:
  print(response)
```

```python
responses=model.batch(["why are macaws majorely red?",
                      "why are penguins only seen in cold areas?",
                      "what is quantum computing"],
                      config={'max_concurrency':5,})#it say that the llm can actually take 5 concurrent questions at once

for response in responses:
  print(response)
```
