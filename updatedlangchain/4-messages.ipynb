Messages

Messages are the fundamental unit of context for models in LangChain. They represent the input and output of models, carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM. Messages are objects that contain:
- Role - Identifies the message type (e.g. system, user)
- Content - Represents the actual content of the message (like text, images, audio, documents, etc.)
- Metadata - Optional fields such as response information, message IDs, and token usage
LangChain provides a standard message type that works across all model providers, ensuring consistent behavior regardless of the model being called.
Text Prompts
Text prompts are strings of text that are passed to the model as input. They are the simplest form of prompting and are commonly used when you want to ask the model a question or give it an instruction.

```python
#ai _msg is used to track the response sent by the ai

#creating an ai message manually instead of waiting for the response from the ai(just to demonstrate how conversation history is mainatained in an ai)
ai_msg=AIMessage("id be happy to help you with this conversation")

#add it to conversation history

message=[
  SystemMessage("you are a helpful ai"),
  HumanMessage("can you help me"),
  ai_msg,
  HumanMessage("great what is 2+2")
]

response = model.invoke(message)
print(response.content)
```

# Tool Message in LangChain

A **ToolMessage** represents the output returned by a tool after it has been executed.

When an LLM requests a tool, the application executes the corresponding Python function. The result returned by that function is represented as a ToolMessage and added to the conversation history.

### How does it work?

1. **HumanMessage:** Contains the user's query.
2. **AIMessage:** The LLM processes the query and generates a tool-call request.
3. **Tool Execution:** The application executes the requested Python function.
4. **ToolMessage:** Contains the result returned by the executed tool.
5. **AIMessage:** The LLM receives the tool result and generates the final response.

### Example

Suppose the user asks:

"What is the weather in Bangalore?"

The LLM requests the `get_weather` tool.

The tool executes and returns:

`The weather in Bangalore is sunny.`

LangChain represents this output as a ToolMessage.

### Important distinction

| Message Type | Responsibility |
|---|---|
| SystemMessage | Provides instructions to the LLM. |
| HumanMessage | Contains user input. |
| AIMessage | Contains the LLM's response or tool-call request. |
| ToolMessage | Contains the output returned by an executed tool. |

### Execution Flow

**HumanMessage → AIMessage (Tool Call) → Tool Execution → ToolMessage (Tool Result) → AIMessage (Final Response)**

**Remember:** AIMessage requests the tool, whereas ToolMessage carries the result returned by the tool.

```python
from langchain.messages import ToolMessage
# After a model makes a tool call
# (Here, we demonstrate manually creating the messages for brevity)

ai_message = AIMessage(
    content=[],
    tool_calls=[
        {
            "name": "get_weather",
            "args": {"location": "San Francisco"},
            "id": "call_123"
        }
    ]
)

# Execute tool and create result message
weather_result = "Sunny, 72°F"

tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"  # Must match the call ID
)

# Continue conversation
messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_message,  # Model's tool call
    tool_message,  # Tool execution result
]

# Send messages back to the model
response = model.invoke(messages)

# Display the model's response
print(response.content)

tool_message
```
